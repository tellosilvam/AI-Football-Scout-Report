### Pandas Scraping HTML

```python
import pandas as pd
import requests
```

```python
# fbref table link
url_df = 'https://fbref.com/en/comps/Big5/gca/players/Big-5-European-Leagues-Stats#stats_gca'

# Disable SSL verification and fetch the HTML content
response = requests.get(url_df, verify=False)
df = pd.read_html(response.text)[0]
df.head()
```

```python
df.columns = [' '.join(col).strip() for col in df.columns]

df = df.reset_index(drop=True)
df.head()
```

```python
# creating a list with new names
new_columns = []
for col in df.columns:
  if 'level_0' in col:
      new_col = col.split()[-1]  # takes the last name
  else:
      new_col = col
  new_columns.append(new_col)

# rename columns
df.columns = new_columns
df = df.fillna(0)

df.head()
```

```python
df.dropna(subset=['Rk'], inplace=True)
```

```python
url = 'https://fbref.com/en/squads/19538871/Manchester-United-Stats'
response = requests.get(url)
df = pd.read_html(response.text, attrs={"id": "stats_standard_9"})[0]
```

```python
df.columns = df.columns.get_level_values(1)
df.columns = df.columns.str.replace(' ', '_')
df.head()
```
